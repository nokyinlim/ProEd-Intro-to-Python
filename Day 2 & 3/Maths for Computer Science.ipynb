#**Standard Deviation**

Next we are going to calculate the Standard Deviation of a data set.


$ SD = \sqrt{\frac{\sum{|x-\mu|^{2}}}{N}}$


````python
from math import sqrt
#Calculate the mean

mean = np.mean(x)

#Calculate SD
s: float = 0
for i in x:
    s += ((i - mean) ** 2)

standardDeviation = sqrt(s / len(x))

print(f"{standardDeviation = }")
````

#**Trendlines**


**Plotting**

We're going to first plot the data so we can see it, don't alter this code just run it!

````python


#We're going to use polyfit to calculate our trendline equation
#polyfit needs 3 entry (x data, y data, the degree of the function you want
#So we're going to look at a first degree polynomial (the highest power of x is one)




plt.plot(x, y, linestyle='', marker='o', color="#ff0000")
pf = np.polyfit(x, y, 1)
pf_y: list[float] = [pf[0] * i + pf[1] for i in range(0, 100, 1)]
print(f"line equation y = {pf[0]:.2f}x + {pf[1]:.2f}")

#add trendline to plot
plt.plot(pf_y, color="#0000ff", linewidth=4)
````

We can see here our trendline looks like a pretty good fit for out data but we can see our data isn't a perfectly aligned.

#**Correlation**

Next we're going to calculate the correlation of our x and y data.
We can see from the trendline the data gives a linear trend but there are some outlying points.

Correlation_1.avif




````python
#Calcualting Correlation
#Look into the Numpy module and how to use correlation


l: list[list[float]] = np.corrcoef(y, pf_y)

l
correlation = l[0][1]
print(f"{correlation = }")
````

````python
import pandas as pd
import numpy as np
x2 = np.random.randint(0, 10, 100)
y2 = np.arange(0, 100, 1)
plt.plot(x2, y2, 'o')
````

````python
#Find the correlation of the data set

p = np.polyfit(x2, y2, 1)

cy2 = [i * p[0] + p[1] for i in x2]

plt.plot(x2, y2, 'o')

plt.plot(x2, cy2)


l: list[list[float]] = np.corrcoef(y2, cy2)


correlation = l[0][1]
print(f"{correlation = :.5f} = {correlation*100:.2f}%")
````
